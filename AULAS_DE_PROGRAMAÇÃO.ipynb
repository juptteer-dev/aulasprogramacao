```python
# Lista de notas dos estudantes
notas = [7.5, 8.0, 6.5, 9.0, 7.0]

# Cria uma função "calcular_media"
def calcular_media(notas):
    # Cria uma variável "total" a gente coloca uma função para somar as notas.
    total = sum(notas)
    # Cria outra variável para calcular a média que é a soma das notas "total" divido pela quantidade de notas "len(notas)"
    media = total / len(notas)
    # Devolve a média calculada
    return media

# Função lambda para arredondar a média para duas casas decimais
arredondar_media = lambda media: round(media, 2)

# Calcular a média
media = calcular_media(notas)
media_arredondada = arredondar_media(media)

# Verificar se os estudantes foram aprovados
situacao = "Aprovado" if media_arredondada >= 7 else "Reprovado"

# Resultados
print(notas)
print(media_arredondada)
print(situacao)
```

```python
# Função def pelo usuário!

def calcular_desconto(valor, percentual):
  """Calcula o valor final adicionando um desconto percentual."""
  if percentual < 0 or percentual > 100:
    return None #Marcação Inválida
  desconto = valor * (percentual / 100)
  return valor - desconto

def registro_venda (produto, valor_final):
  """Imprime o registro de uma venda."""
  print (f">>Venda registrada: {produto}, por R$ {valor_final:.2f}")

# Função lambda para arredondar
arredondar = lambda v: round(v, 2)
```

```python
# Variavéis principais do PDV (Produto De Venda)
total_vendas = 0
qtd_vendas = 0
```

```python
# Loop principal do caixa. MINHALOJAPVD
while True:
  print ("n----- minha loja - pvd -----n")
  print ("1 - Registrar Venda")
  print ("2 - Ver relatório do dia")
  print ("3 - Encerrar Caixa")
  opcao = input("Escolha uma opção: ")

  if opcao == "3":
    print ("Encerrando o caixa...")
    break
  elif opcao == "1":
    produto = input("Nome do Produto: ")
    valor = float(input("Valor do Produto: R$ "))
    desconto = float(input("Desconto (em %): "))
    valor_final = calcular_desconto(valor, desconto)

    if valor_final is None:
      print ("Desconto Inválido!")

    else:
      valor_final = arredondar(valor_final)
    registro_venda(produto, valor_final)
    total_vendas = total_vendas + valor_final
    qtd_vendas = qtd_vendas + 1

  elif opcao == "2":
    print (f"\n Quantidade de Vendas: {qtd_vendas}")
    print (f"Total Faturado: R$ {arredondar(total_vendas)}")

  else:
    print ("Opção Inválida!")
```

```python
# Serve para além de digitar numpy toda vez escrever np
import numpy as np

# Cria uma lista [] e cada participante dessa lista tem um dicionário {}
participantes = [
{"nome": "Gabispeixe", "localizacao": "EUA", "afiliacao": "IFCE", "interesses": ["Física", "Astronomia"]},
{"nome": "Pê", "localizacao": "Brasil", "afiliacao": "UFC","interesses": ["Biologia", "Astronomia"]},
{"nome": "Larissa", "localizacao": "Colombia", "afiliacao": "ANHANGUERA", "interesses": ["Química", "Engenharia"]}
]
print(f"Lista de participantes: {participantes}")

print ("----------------------------------------")

# Usando sets para identificar diferentes regiões dos participantes e não aceitando elementos repetidos
regioes = set(participante["localizacao"] for participante in participantes)

# Usando um dicionário para categorizar afiliações
afiliacoes = {}

for participante in participantes:
    afiliacao = participante["afiliacao"]
    if afiliacao not in afiliacoes:
        afiliacoes[afiliacao] = []
    afiliacoes[afiliacao].append(participante["nome"])

# Usando NumPy para analisar áreas de interesse
areas_de_interesse = np.array([interesse for participante in participantes for interesse in participante["interesses"]])
interesses_unicos, contagem = np.unique(areas_de_interesse, return_counts=True)
area_mais_popular = interesses_unicos[np.argmax(contagem)]

# Resultados
print("Regiões dos participantes:", regioes)

print ("----------------------------------------")

print("Afiliações dos participantes:")
for afiliacao, nomes in afiliacoes.items():
    print(f"{afiliacao}: {', '.join(nomes)}")

print ("----------------------------------------")

print("Área de interesse mais popular:", area_mais_popular)
```

```python
# Criei uma classe "pai", defini as caracteristicas que todo animal tem e defini um barulho
class Animal:
  def __init__(self, nome, especie):
    self.nome = nome
    self.especie = especie

# Pass significa que o animal irá herdar est função e terá seu próprio barulho
  def barulho(self):
    pass

# Está é um exemplo de classe "filha" uma classe derivada da classe pai
class Cachorrro(Animal):
  def barulho(self):
    return "Latido"

animal1 = Cachorrro("Bolt", "Pastor-branco-americano")
print (f"Essa é o {animal1.nome} e a espécie dele é {animal1.especie}")
animal1.barulho()

print ("\n-------------------------------------------\n")

# Crie uma classe "pai" e adicionei a velocidade e coloquei 0 pois todo automóvel começa parado.
class Automovel:
  def __init__(self, marca, modelo, ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano
    self.velocidade = 0

  def acelerar(self, incremento):
    self.velocidade += incremento

  def frear(self, incremento):
    self.velocidade -= incremento

  def status(self):
    return f"Marca: {self.marca}, Modelo: {self.modelo}, Ano: {self.ano}, Velocidade: {self.velocidade}"

# Agora as classes filhas.
class Moto(Automovel):
  def __init__(self, marca, modelo, ano, cilindradas):
# Super funciona para separar o que é da classe pai (automovel) e o que é da classe filha (moto)
      super().__init__(marca, modelo, ano)
      self.cilindradas = cilindradas

  def status(self):
    return f"Marca: {self.marca}, Modelo: {self.modelo}, Ano: {self.ano}, Velocidade: {self.velocidade}, Cilindradas: {self.cilindradas}"

class Carro(Automovel):
  def __init__(self, marca, modelo, ano, potencia):
      super().__init__(marca, modelo, ano)
      self.potencia = potencia

# Adicionei a funcionalidade da função acelerar que é somar a potencia + a velocidade atual
  def acelerar(self, incremento):
    self.velocidade += incremento + self.potencia

moto1 = Moto("Yamaha", "NMAX160", 2022, 155)
carro1 = Carro("Fiat", "Hatchbacks", 2023, 100)

moto1.acelerar(100)
carro1.acelerar(100)

print ("Status da moto:")
print (moto1.status())
print ("\nStatus do carro:")
print (carro1.status())
```

```python
# Importa a biblioteca de gráficos
import matplotlib.pyplot as plt

# Classe-pai Produto, que define nome, preço, categoria e estoque
class Produto:
  def __init__(self, nome, preco, categoria, estoque):
    self.nome = nome
    self.preco = preco
    self.categoria = categoria
    self.estoque = estoque

# Imprime o nome, preço, categoria e o produto
  def __str__(self):
    return f"Nome: {self.nome}, Preço: R$ {self.preco}, Categoria: {self.categoria}, Estoque: {self.estoque}"

# Quando um produto for vendido ele será retirado do estoque
  def vender(self, quantidade):
    if quantidade <= self.estoque:
      self.estoque -= quantidade

# Listas vazias
catalogo = []
categorias = []

# Defini um novo produto para ser adicionado a loja. Append serve para adicionar à lista
def adicionar_produto(nome, preco, categoria, estoque):
  novo_produto = Produto(nome, preco, categoria, estoque)
  catalogo.append(novo_produto)
  categorias.append(categoria)
  print (f"O produto {nome} foi adicionado ao catálogo.")

def retirar_produto(nome):
  for produto in catalogo:
    if produto.nome == nome:
      catalogo.remove(produto)
      print (f"O produto {nome} foi removido do catálogo.")
      return
  print (f"O produto {nome} não foi encontrado no catálogo.")

# Imprime todos os produtos do catálogo
def listar_catalogo():
  print ("\nCatálogo de Produtos:")
  for produto in catalogo:
    print (produto)

# Cadrastamento de produtos e retiro de produtos
adicionar_produto("Mangás", 19.99, "Livros Ocidentais", 100)
adicionar_produto("Manhwa", 35.50, "Livros Ocidentais", 80)
adicionar_produto("Boosters", 27.99, "Cartinhas", 50)
retirar_produto("Mangás")

listar_catalogo()

# Gráfico por categoria
categorias_unicas = list(set(categorias))
categorias_unicas.sort()

# Contagem de produtos em categoria
contagem_categorias = [categorias.count(categoria) for categoria in categorias_unicas]

print ("\n-----------------------------------\n")

plt.bar(categorias_unicas, contagem_categorias)
plt.xlabel("Categorias")
plt.ylabel("Quantidade")
plt.title("Quantidade de Produtos por Categoria")
plt.show()
```

# Sistemas de gerenciamento de Livros

```python
import matplotlib.pyplot as plt

class Livros:
  def __init__(self, titulo, autor, genero, quantidade_disponivel):
    self.titulo = titulo
    self.autor = autor
    self.genero = genero
    self.quantidade_disponivel = quantidade_disponivel

  def __str__(self):
    return f"Título: {self.titulo}, Autor: {self.autor}, Gênero: {self.genero}, Unidades Disponíveis: {self.quantidade_disponivel}"

estantes = []
genero = []

def adicionar_livro(titulo, autor, genero, quantidade_disponivel):
  novo_livro = Livros(titulo, autor, genero, quantidade_disponivel)
  estantes.append(novo_livro)

print ("\n---- Sistemas de gerenciamento de Livros ----\n")

def listar_estante():
  print ("\nEstante de Livros:")
  for livro in estantes:
    print (livro)

#Cadastro de novos livros a estante:
adicionar_livro ("Garoto de Ipanema", "Manu Barbosa", "Romance", 15)
adicionar_livro ("Divina Comédia", "Dante Alighieri", "Poemas épicos", 33)
adicionar_livro ("O Rei de Amerelo", "Robert W. Chambers", "Contos de terror", 12)
adicionar_livro ("O Médico e o Monstro", "Robert Louis Stevenson", "Contos de terror", 20)
adicionar_livro ("Orgulho e Preconceito", "Jane Austen", "Romance", 10)
adicionar_livro ("Drácula", "Bram Stoker", "Contos de terror", 37)
adicionar_livro ("O Chamado de Cthulhu", "H. P. Lovecraft", "Contos de terror", 12)
adicionar_livro ("Frankenstein", "Mary Shelley", "Contos de terror", 45)
adicionar_livro ("Jane Eyre", "Charlotte Brontë", "Romance", 23)
adicionar_livro ("Dom Casmurro", "Machado de Assis", "Romance", 8)
adicionar_livro ("A Culpa é das Estrelas", "John Green", "Romance", 41)
adicionar_livro ("A Odisseia", "Homero", "Poemas épicos", 16)
adicionar_livro ("A Ilíada", "Homero", "Poemas épicos", 49)

print(f"\nForam cadastrados {len(estantes)} livros.")

print ("\n-----------------------------------")

listar_estante()

print ("\n-----------------------------------\n")

buscar_livro = input(f"Digite o título do livro que deseja buscar: ").lower()
if len(buscar_livro) < 3:
    print("Digite pelo menos 3 caracteres para realizar a busca.")
else:
    resultado_busca = [
        livro for livro in estantes
        if buscar_livro in livro.titulo.lower()
    ]

if resultado_busca:
  print("\nLivros encontrados:")
  for livro in resultado_busca:
    print(f"Achamos o livro:\n {livro}")
else:
  print("Nenhum livro corresponde à sua busca.")

print ("\n-----------------------------------\n")

genero_unicas = list(set(livro.genero for livro in estantes))
genero_unicas.sort()

contagem_genero = [
    sum(1 for livro in estantes if livro.genero == genero)
    for genero in genero_unicas
]

plt.bar(genero_unicas, contagem_genero)
plt.title("Quantidade de Livros por Gênero")
plt.xlabel("Gênero")
plt.ylabel("Quantidade")
plt.show()
```
