```python
import unittest as ut

def aplicar_desconto(total, percentual):
  valor_descontado = percentual / 100 * total
  resultado = total - valor_descontado
  return resultado

class TestAplicarDesconto(ut.TestCase):

  def test_aplicar_desconto(self):
    self.assertEqual(aplicar_desconto(100, 10), 90.0)

  def test_aplicar_desconto_zero(self):
    self.assertEqual(aplicar_desconto(200, 0), 200.0)

ut.main(argv=[''], exit=False, verbosity=2)
```
