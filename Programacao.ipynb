```python
# TF - SUPERVISIONADO
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dados rotulados (entrada + saída esperada)
x_train = tf.constant([[1.0], [2.0], [3.0], [4.0]])
y_train = tf.constant([[2.0], [4.0], [6.0], [8.0]])
```

```python
model = Sequential()
model.add(Dense(units=1, input_shape=(1,)))
model.compile(optimizer='sgd', loss='mean_squared_error')
history = model.fit(x_train, y_train, epochs=1000, verbose=0)
```

```python
X_new = tf.constant([[5.0]])
prediction = model.predict(X_new, verbose=0)
print (f"Predição para x=5: {prediction[0][0]:.2f} (esperado: 10)")
```

```python
# Curva de aprendizado
plt.plot(history.history['loss'])
plt.title('Loss ao longo do treinamento')
plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.show()
```

```python
#2. Treinamento NÃO SUPERVISIONADO - autoencoder mínimo
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model

# Dados sem rótulo (só X)
X_unsupervised = tf.constant([[1.0, 2.0], [2.0, 3.0], [3.0, 4.0], [4.0, 5.0]])

# Autoencoder: encode comprime para 1 dimensão; decode reconstrói para 2
input_layer = Input(shape=(2,))
encoded = Dense(units=1)(input_layer)
decoded = Dense(units=2)(encoded)
autoencoder = Model(inputs=input_layer, outputs=decoded)
autoencoder.compile(optimizer='adam', loss='mean_squared_error')

# Treina tentando reconstruir a própria entrada
autoencoder.fit(X_unsupervised, X_unsupervised, epochs=1000, verbose=0)

prediction = autoencoder.predict(X_unsupervised, verbose=0)
print("Original:", X_unsupervised.numpy())
print("Reconstruido:", prediction)
```
